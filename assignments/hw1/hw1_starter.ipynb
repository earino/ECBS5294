### ⚠️ Understand the Grain Before You Query

**The single most important question about any table: what does one row represent?**

In this dataset, **one row = one invoice line item** — a single product on a single invoice. One customer checkout (an *invoice*, identified by the `Invoice` column) usually contains **many** line items: a basket with 12 different products produces 12 rows sharing one invoice number.

This means:
- `COUNT(*)` counts **line items**, not purchases
- `COUNT(DISTINCT Invoice)` counts **transactions** (checkouts)
- `COUNT(DISTINCT "Customer ID")` counts **customers**

Mixing these up silently inflates business metrics ~18x in this data. Run the next cell and keep these three numbers in mind for the whole assignment.

````python
# The three grains of this dataset — know these numbers!
grain = con.execute("""
    SELECT
        COUNT(*)                        AS line_items,
        COUNT(DISTINCT Invoice)         AS invoices,
        COUNT(DISTINCT "Customer ID")   AS customers
    FROM retail
""").df()

# Sanity checks: the grain is line items, and invoices contain many of them
assert grain['line_items'][0] > grain['invoices'][0], "Expected many line items per invoice!"
assert grain['invoices'][0] > grain['customers'][0], "Expected multiple invoices per customer!"

print(f"Each invoice has on average {grain['line_items'][0] / grain['invoices'][0]:.1f} line items.")
grain
````

**Good!** Now you know:
- Total row count — and that a *row is a line item, not a transaction*
- Which columns have NULLs (Customer ID and Description)
- Date range covered

Whenever a question says *transactions*, ask yourself: does it want line items (`COUNT(*)`) or invoices (`COUNT(DISTINCT Invoice)`)? Keep this in mind as you write queries!
